```python
from pathlib import Path
import pathlib
#'C://dev//dolphin_initial_testing//DOLPHIN//xgb_lbd_hc_loo_9_1'
def sumarry_file_to_df(diagnosis_directory_path):
    directory_path = Path(diagnosis_directory_path)
    metrics_paths_list = list(Path(directory_path).rglob('metrics.json'))
    metrics_paths_list
    rows = []
    for metrics_path in metrics_paths_list:
        with open(metrics_path,"r") as f:
            data = json.load(f)
        featureset_row = metrics_path.parent.parent.name
        # Convert top-level metrics
        rows.append({
            "feature set": featureset_row, 
            "balanced_accuracy": data["balanced_accuracy"],
            "mcc": data["mcc"],
            "roc_auc": data["roc_auc"],
            "sensitivity": data["sensitivity"],
            "specificity": data["specificity"]
        })
    df_summary = pd.DataFrame(rows)
    return df_summary 


```

```python

from pathlib import Path
import pathlib
#'C://dev//dolphin_initial_testing//DOLPHIN//xgb_lbd_hc_loo_9_1'
def multitask_sumarry_file_to_df(diagnosis_directory_path):
    directory_path = Path(diagnosis_directory_path)
    metrics_paths_list = list(Path(directory_path).rglob('metrics.json'))
    metrics_paths_list
    rows = []
    for metrics_path in metrics_paths_list:
        with open(metrics_path,"r") as f:
            data = json.load(f)
        featureset_row = metrics_path.parent.parent.name
        task_row = metrics_path.parent.name
        # Convert top-level metrics
        rows.append({
            "feature set": featureset_row, 
            "task": task_row, 
            "balanced_accuracy": data["balanced_accuracy"],
            "mcc": data["mcc"],
            "roc_auc": data["roc_auc"],
            "sensitivity": data["sensitivity"],
            "specificity": data["specificity"]
        })
    df_summary = pd.DataFrame(rows)
    return df_summary 
```
